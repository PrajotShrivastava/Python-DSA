```python
s1="act"
s2="cat"
mp1={}
mp2={}
for ch in s1:
    if ch in mp1:
        mp1[ch]+=1
    else:
        mp1[ch]=1
for ch in s2:
    if ch in mp2:
        mp2[ch]+=1
    else:
        mp2[ch]=1
if mp1 == mp2:
    print("True")
else:
    print("False")
```
